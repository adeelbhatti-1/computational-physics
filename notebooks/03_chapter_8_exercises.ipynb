```python
#part (a)

import numpy as np
import matplotlib.pyplot as plt

def square_wave(t):
    """Square wave input: 1 if floor(2t) is even, -1 otherwise."""
    return 1 if np.floor(2 * t) % 2 == 0 else -1

def dVout_dt(Vout, t, RC):
    """Differential equation for the low-pass filter."""
    return (square_wave(t) - Vout) / RC

def rk4_solve(RC, t_end=10, h=0.01, Vout0=0):
    """Solves the ODE using RK4 method."""
    t_points = np.arange(0, t_end, h)
    Vout_points = []
    Vout = Vout0

    for t in t_points:
        Vout_points.append(Vout)
        # RK4 method
        k1 = h * dVout_dt(Vout, t, RC)
        k2 = h * dVout_dt(Vout + 0.5 * k1, t + 0.5 * h, RC)
        k3 = h * dVout_dt(Vout + 0.5 * k2, t + 0.5 * h, RC)
        k4 = h * dVout_dt(Vout + k3, t + h, RC)
        Vout += (k1 + 2 * k2 + 2 * k3 + k4) / 6

    return t_points, Vout_points

# Parameters
RC_values = [0.01, 0.1, 1]
h = 0.001  # Step size (smaller = more accurate but slower)
t_end = 10

# Solve and plot for each RC
plt.figure(figsize=(10, 6))
for RC in RC_values:
    t, Vout = rk4_solve(RC, t_end, h)
    plt.plot(t, Vout, label=f"RC = {RC}")

plt.xlabel("Time (t)")
plt.ylabel("Output Voltage (Vout)")
plt.title("Low-Pass Filter Response to Square Wave Input")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def lorenz_equations(state, t, sigma, rho, beta):
    """Lorenz equations: dx/dt, dy/dt, dz/dt."""
    x, y, z = state
    dxdt = sigma * (y - x)
    dydt = rho * x - y - x * z
    dzdt = x * y - beta * z
    return np.array([dxdt, dydt, dzdt])

def rk4_solve(func, initial_state, t_points, sigma, rho, beta):
    """Solves the ODE system using RK4."""
    # Ensure state is a float array to prevent casting errors
    state = initial_state.copy().astype(float)
    solution = []
    for t in t_points:
        solution.append(state.copy())
        # RK4 method
        k1 = h * func(state, t, sigma, rho, beta)
        k2 = h * func(state + 0.5 * k1, t + 0.5 * h, sigma, rho, beta)
        k3 = h * func(state + 0.5 * k2, t + 0.5 * h, sigma, rho, beta)
        k4 = h * func(state + k3, t + h, sigma, rho, beta)
        state += (k1 + 2 * k2 + 2 * k3 + k4) / 6
    return np.array(solution)

# Parameters
sigma, rho, beta = 10, 28, 8/3  # Classic chaotic parameters
# Explicitly set the data type to float when creating the initial state
initial_state = np.array([0, 1, 0], dtype=float)  # (x, y, z) at t=0

# Time settings
t_start, t_end = 0, 50
h = 0.01  # Step size
t_points = np.arange(t_start, t_end, h)

# Solve the system
solution = rk4_solve(lorenz_equations, initial_state, t_points, sigma, rho, beta)

# Plot y(t)
plt.figure(figsize=(10, 5))
plt.plot(t_points, solution[:, 1], 'b', linewidth=0.5)
plt.xlabel("Time (t)")
plt.ylabel("y(t)")
plt.title("Lorenz System: y(t) vs. Time (Chaotic Behavior)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the harmonic_oscillator and anharmonic_oscillator functions again for completeness
def harmonic_oscillator(state, t, omega):
    x, v = state
    dxdt = v
    dvdt = -omega**2 * x
    return np.array([dxdt, dvdt])

def anharmonic_oscillator(state, t, omega):
    x, v = state
    dxdt = v
    dvdt = -omega**2 * x**3
    return np.array([dxdt, dvdt])

# Define the lorenz_equations function again for completeness
def lorenz_equations(state, t, sigma, rho, beta):
    """Lorenz equations: dx/dt, dy/dt, dz/dt."""
    x, y, z = state
    dxdt = sigma * (y - x)
    dydt = rho * x - y - x * z
    dzdt = x * y - beta * z
    return np.array([dxdt, dydt, dzdt])


# Modified rk4_solve to accept arbitrary additional arguments for the function func
def rk4_solve(func, initial_state, t_points, *args): # Added *args here
    """Solves the ODE system using RK4."""
    # Ensure state is a float array to prevent casting errors
    state = initial_state.copy().astype(float)
    solution = []
    for t in t_points:
        solution.append(state.copy())
        h = t_points[1] - t_points[0]  # Step size - recalculate for each loop to be safe with non-uniform t_points
        # RK4 method - pass *args to func
        k1 = h * func(state, t, *args)
        k2 = h * func(state + 0.5 * k1, t + 0.5 * h, *args)
        k3 = h * func(state + 0.5 * k2, t + 0.5 * h, *args)
        k4 = h * func(state + k3, t + h, *args)
        state += (k1 + 2*k2 + 2*k3 + k4) / 6
    return np.array(solution)

# Define the vanderpol_oscillator function
def vanderpol_oscillator(state, t, mu, omega):
    x, v = state
    dxdt = v
    dvdt = mu * (1 - x**2) * v - omega**2 * x
    return np.array([dxdt, dvdt])

# Parameters
mu_values = [1, 2, 4]
omega = 1.0
# initial_state needs to be defined here or globally before this cell
# Assuming initial_state = np.array([1.0, 0.0]) from the previous block
initial_state = np.array([1.0, 0.0])
t_start, t_end = 0, 20
t_points = np.linspace(t_start, t_end, 10000)  # Using linspace is generally safer for fixed step size

# Solve and plot phase space
plt.figure(figsize=(10, 8))
for mu in mu_values:
    # Pass mu and omega as separate arguments after t_points
    solution = rk4_solve(vanderpol_oscillator, initial_state, t_points, mu, omega)
    plt.plot(solution[:, 0], solution[:, 1], label=f'μ={mu}', linewidth=0.5)
plt.xlabel("Position (x)")
plt.ylabel("Velocity (v)")
plt.title("Van der Pol Oscillator Phase Space")
plt.grid(True)
plt.legend()
plt.show()
```
