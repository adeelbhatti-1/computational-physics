```python
import numpy as np
import matplotlib.pyplot as plt

def mandelbrot_set(width=800, height=800, xmin=-2, xmax=1, ymin=-1.5, ymax=1.5, max_iter=100):
    x = np.linspace(xmin, xmax, width)
    y = np.linspace(ymin, ymax, height)
    X, Y = np.meshgrid(x, y)
    C = X + 1j * Y
    Z = np.zeros_like(C, dtype=np.complex128)
    mandelbrot = np.full(C.shape, True, dtype=bool)

    for i in range(max_iter):
        Z[mandelbrot] = Z[mandelbrot]**2 + C[mandelbrot]
        mandelbrot[np.abs(Z) > 2] = False

    return mandelbrot, x, y

def plot_mandelbrot_with_c(c, max_iter=50):
    mandelbrot, x, y = mandelbrot_set()

    # Compute the trajectory of given c
    z = 0
    trajectory = []
    for _ in range(max_iter):
        z = z**2 + c
        trajectory.append(z)
        if abs(z) > 2:
            break

    # Convert trajectory to real and imaginary parts
    x_traj, y_traj = np.real(trajectory), np.imag(trajectory)

    # Plot Mandelbrot set
    plt.figure(figsize=(8, 8))
    plt.imshow(mandelbrot, extent=[-2, 1, -1.5, 1.5], cmap='magma', origin='lower')
    plt.colorbar(label='Iterations')

    # Plot trajectory of c
    plt.plot(x_traj, y_traj, 'wo-', markersize=4, label=f'Trajectory for c = {c}')
    plt.scatter(np.real(c), np.imag(c), color='red', marker='x', s=100, label='c')

    plt.title("Mandelbrot Set with Arbitrary c")
    plt.xlabel("Real Axis")
    plt.ylabel("Imaginary Axis")
    plt.legend()
    plt.show()

# Example: Change this value to any complex number
c_value = -0.3 + 0.0j
plot_mandelbrot_with_c(c_value)
```
