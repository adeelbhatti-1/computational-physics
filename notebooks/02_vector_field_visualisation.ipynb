```python
import numpy as np
import matplotlib.pyplot as plt

def plot_vector_field(vector_function, x_range, y_range, density=20):
    """
    Plots a 2D vector field for a given vector function.

    Parameters:
    - vector_function: A function that takes two arguments (x, y) and returns a tuple (u, v).
    - x_range: A tuple specifying the range of x-axis values (x_min, x_max).
    - y_range: A tuple specifying the range of y-axis values (y_min, y_max).
    - density: Number of points along each axis for the vector grid (default: 20).
    """
    # Generate grid points
    x = np.linspace(x_range[0], x_range[1], density)
    y = np.linspace(y_range[0], y_range[1], density)
    X, Y = np.meshgrid(x, y)

    # Compute vector field components
    U, V = np.zeros_like(X), np.zeros_like(Y)
    for i in range(X.shape[0]):
        for j in range(X.shape[1]):
            U[i, j], V[i, j] = vector_function(X[i, j], Y[i, j])

    # Plot the vector field
    plt.figure(figsize=(8, 6))
    plt.quiver(X, Y, U, V, color='blue')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('2D Vector Field')
    plt.axis('equal')
    plt.grid(True)
    plt.show()

# Example usage
def example_vector_function(x, y):
    return 2*x*np.exp(x**2+y**2), -2*y*np.exp(x**2+y**2)  # Rotational field

plot_vector_field(example_vector_function, x_range=(-5, 5), y_range=(-5, 5))
```
